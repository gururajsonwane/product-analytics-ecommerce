```python
OUT = r"C:\datasets\events_sample_oct2019.csv"

total_rows = 0
kept_rows = 0
first_chunk = True
```

```python
chunks = pd.read_csv(SRC, chunksize=500_000)

for chunk in chunks:
    total_rows += len(chunk)
    sample = chunk[chunk["user_id"] % 10 == 0]
    kept_rows += len(sample)
    sample.to_csv(OUT, mode="w" if first_chunk else "a", header=first_chunk, index=False)
    first_chunk = False
    print(total_rows, kept_rows)

print("DONE", total_rows, kept_rows)
```
